```python
# Sales Data Calculator

# Step 1: Create a list of sales values
sales = [1200, 2500, 1800, 3200, 1500, 4000, 2800]

# Step 2: Calculate total sales
total_sales = sum(sales)

# Step 3: Calculate the number of sales
number_of_sales = len(sales)

# Step 4: Calculate average sales
average_sales = total_sales / number_of_sales

# Step 5: Find the highest sale
highest_sale = max(sales)

# Step 6: Find the lowest sale
lowest_sale = min(sales)

# Step 7: Display sales data
print("SALES DATA")
print(sales)

# Step 8: Display the analysis results
print("\nSALES ANALYSIS")

print("Number of sales:", number_of_sales)

print(f"Total sales: ₹{total_sales:,.2f}")

print(f"Average sale: ₹{average_sales:,.2f}")

print(f"Highest sale: ₹{highest_sale:,.2f}")

print(f"Lowest sale: ₹{lowest_sale:,.2f}")

# Step 9: Explain the results
print("\nRESULT EXPLANATION")

print(f"The total value of {number_of_sales} sales is ₹{total_sales:,.2f}.")

print(f"The average sale value is ₹{average_sales:,.2f}.")

print(f"The highest sale is ₹{highest_sale:,.2f}.")

print(f"The lowest sale is ₹{lowest_sale:,.2f}.")
```
